# 🖨️ The Python `print()` Function

The `print()` function is a built-in function used to output text, variables, or data structures directly to the console or screen.

## ⚙️ Core Parameters & Syntax

```python
print(value1, value2, ..., sep=' ', end='\n', file=sys.stdout, flush=False)
```

* **`values`** - Any object(s) you want to print. Multiple objects are separated by commas.
* **`sep`** - Specifies how to separate multiple objects. *(Default is a single space `' '`)*
* **`end`** - Specifies what to print at the very end of the line. *(Default is a newline `'\n'`)*
* **`file`** - An object with a write method where the output goes. *(Default is the screen / `sys.stdout`)*

---

## 💡 Common Examples

```python
# 1. Basic Printing
print("Hello, World!")

# 2. Printing Multiple Items with a Custom Separator
print("Python", "Data", "Science", sep=" ➔ ")
# Output: Python ➔ Data ➔ Science

# 3. Changing the End Character (Prevents moving to a new line)
print("Hello", end=" ")
print("World!")
# Output: Hello World!
```
