# 🐍 Python Data Types

Python supports **3 categories** of data types:

* **Basic Types** - `integer`, `float`, `complex`, `boolean`, and `string`
* **Container Types** - `List`, `Tuples`, `Sets`, and `Dictionary`
* **User-defined Types** - `Class`


```python
# Integer
print(6)
print(1e309)
```

```python
# Float
print(6.5)
print(1.7e309)
```

```python
# Boolean
print(True)
print(False)
```

```python
# Complex
print(4+5j)
```

```python
# String
print('Pakistan')
print("Pakistan")
print("""Pakistan""")
```

```python
# List
print([1, 2, 3, 4, 5])
```

```python
# Tuple
print((1,2,3,4,5))
```

```python
# Sets
print({1, 2, 3, 4, 5})
```

```python
# Dictionary
print({"Name":"Haseeb", "Age":22, "Gender":"Male"})
```
