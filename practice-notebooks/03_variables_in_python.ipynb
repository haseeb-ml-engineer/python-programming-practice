# 📦 Variables in Python

A **variable** is a named storage location in a computer's memory used to hold data that can be referenced and manipulated throughout a program. Think of a variable as a labeled box that holds a value.

---

## ⚙️ Core Characteristics

* **Dynamic Typing:** You do not need to declare the data type explicitly. Python automatically detects the type based on the value assigned.
* **Re-assignment:** A variable can change its value and even its data type at any point during execution.
* **Reference Mechanics:** In Python, variables point to the memory address where an object lives, rather than storing the raw value directly in the variable name itself.

---

## 📐 Variable Naming Rules & Conventions

* **Allowed:** Must start with a letter or an underscore (`_`), followed by letters, numbers, or underscores (e.g., `user_age`, `_temp`).
* **Forbidden:** Cannot start with a number (e.g., `1st_place`) or contain special characters like `-`, `$`, or spaces.
* **Case-Sensitive:** `age`, `Age`, and `AGE` are treated as three completely separate variables.
* **Naming Style:** The PEP 8 style guide recommends using **snake_case** for variable names (lowercase words separated by underscores).

---

```python
# 1. Declaration and Dynamic Typing
```

# 🔄 Dynamic Binding

**Dynamic Binding** means that Python links variable names to objects and checks their types **at runtime** rather than during compilation. 

Variables are just pointers to memory locations. You can freely re-bind a variable name to an entirely different data type at any point in your code.

---

### 💡 Example

```python
x = 5
print(type(x))  # Output: <class 'int'>

# Dynamically re-bound at runtime
x = "Python"
print(type(x))  # Output: <class 'str'>
```

> **Key Rule:** In Python, variable names have no fixed types; only the objects they point to have types.


````python
a=5; b=6; c=7
print(a)
print(b)
print(c)
````

````python
a,b,c = 4,5,6
print(a)
print(b)
print(c)
````

````python
a= b= c=6
print(a)
print(b)
print(c)
````
