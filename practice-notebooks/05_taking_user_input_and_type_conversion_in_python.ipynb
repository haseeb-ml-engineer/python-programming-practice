# 📥 User Input and Type Conversion

Python allows you to interact with users by capturing their input, which always arrives as text and can be converted into numerical formats when needed.

---

### 1. Taking User Input via `input()`
The built-in **`input()`** function pauses program execution and waits for the user to type something. 

> ⚠️ **The Golden Rule:** The `input()` function **always returns data as a string (`str`)**, even if the user types a raw number.

```python
# Captures input as a string
name = input("Enter your name: ") 
print(type(name))  # Output: <class 'str'>
```

---

### 2. Type Conversion (Type Casting)
Because inputs arrive as text, you must explicitly convert them using constructor functions before performing mathematical operations.

* **`int()`** - Converts a string or float into an integer.
* **`float()`** - Converts a string or integer into a decimal number.
* **`str()`** - Converts a number back into text.

---

### 💡 Code Example: Combining Both

```python
# ❌ Broken Approach (Will raise a TypeError)
age = input("Enter your age: ")
# print(age + 1)  # Error: Cannot add an integer to a string

# ✅ Correct Approach (Explicit Type Conversion)
age_input = input("Enter your age: ")
age = int(age_input)  # Convert string to integer

print(f"Next year you will be {age + 1} years old.")
```
