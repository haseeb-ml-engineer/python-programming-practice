# 🔑 Keywords in Python

**Keywords** are reserved words that have predefined meanings and specific purposes in Python. You cannot use these words as names for your variables, functions, or classes because they form the core syntax of the language.

---

### ⚙️ Core Properties
* **Case-Sensitive:** Most keywords are entirely lowercase (e.g., `if`, `for`, `while`), except for three built-in boolean/null values: `True`, `False`, and `None`.
* **Immutable List:** The set of keywords is fixed by Python and can be checked programmatically using the built-in `keyword` module.

---

### 💡 Code Example: How to View All Keywords

Run this code in a Python cell to see the complete, official list for your current version:

```python
import keyword

# Print all reserved keywords
print(keyword.kwlist)
```

> **⚠️ Rule:** Trying to assign a value to a keyword (e.g., `if = 5`) will immediately trigger a `SyntaxError: invalid syntax`.


````python
import keyword

# Print all reserved keywords
print(keyword.kwlist)
````

# 🏷️ Identifiers in Python

An **identifier** is a user-defined name given to program elements like **variables, functions, classes, modules, or packages**. It is how you label things you create in your code.

---

### 📐 Mandatory Rules

To be valid, an identifier must follow these strict rules:

* **Starting Character:** Must start with a letter (`a-z`, `A-Z`) or an underscore (`_`). It **cannot** start with a number.
* **Allowed Elements:** Can contain letters, numbers (`0-9`), and underscores. No spaces or special characters (`@`, `$`, `%`, etc.) are allowed.
* **Reserved Words:** Cannot be any of Python's built-in **keywords** (e.g., you cannot name a variable `def` or `while`).
* **Case-Sensitivity:** `userAge` and `userage` are treated as two completely distinct identifiers.

---

### 💡 Quick Examples

```python
# ✅ Valid Identifiers
age = 25
_user_name = "Alice"
score_2 = 98

# ❌ Invalid Identifiers
2nd_score = 98   # Error: Starts with a number
user-name = "Bob" # Error: Contains a hyphen
if = 10           # Error: Uses a reserved keyword
```
