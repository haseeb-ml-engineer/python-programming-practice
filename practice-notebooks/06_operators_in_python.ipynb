# Python Operators Quick Reference

Operators perform actions on values and variables.

---

### 1. Arithmetic Operators
* `+` **Addition:** `5 + 3` ➡️ `8`
* `-` **Subtraction:** `5 - 3` ➡️ `2`
* `*` **Multiplication:** `5 * 3` ➡️ `15`
* `/` **Division:** `5 / 2` ➡️ `2.5`
* `//` **Floor Division:** `5 // 2` ➡️ `2`
* `%` **Modulus:** `5 % 2` ➡️ `1` (remainder)
* `**` **Exponentiation:** `2 ** 3` ➡️ `8` ($2^3$)

---

### 2. Comparison Operators
* `==` **Equal to:** `5 == 5` ➡️ `True`
* `!=` **Not equal to:** `5 != 3` ➡️ `True`
* `>` **Greater than:** `5 > 3` ➡️ `True`
* `<` **Less than:** `5 < 3` ➡️ `False`
* `>=` **Greater or equal:** `5 >= 5` ➡️ `True`
* `<=` **Less or equal:** `4 <= 3` ➡️ `False`

---

### 3. Logical Operators
* `and` ➡️ Returns `True` if **both** sides are true (e.g., `5 > 3 and 2 > 1`)
* `or` ➡️ Returns `True` if **at least one** side is true (e.g., `5 > 3 or 2 < 1`)
* `not` ➡️ Reverses the boolean result (e.g., `not(5 > 3)` ➡️ `False`)

---

### 4. Assignment Operators
* `=` ➡️ `x = 10`
* `+=` ➡️ `x += 5` (Same as `x = x + 5`)
* `-=` ➡️ `x -= 3` (Same as `x = x - 3`)
* `*=` ➡️ `x *= 2` (Same as `x = x * 2`)

---

### 5. Membership Operators
* `in` ➡️ `'a' in 'apple'` ➡️ `True`
* `not in` ➡️ `'b' not in 'apple'` ➡️ `True`

```python
# 1. ARITHMETIC OPERATORS

a = 10
b = 3

print("--- Arithmetic Operators ---")
print("Addition (+):", a + b)          # 13
print("Subtraction (-):", a - b)       # 7
print("Multiplication (*):", a * b)    # 30
print("Division (/):", a / b)          # 3.3333333333333335
print("Floor Division (//):", a // b)  # 3 (drops decimal)
print("Modulus (%):", a % b)           # 1 (remainder)
print("Exponentiation (**):", a ** b)  # 1000 (10^3)
print()
```

```python
# 2. COMPARISON OPERATORS

x = 5
y = 8

print("--- Comparison Operators ---")
print("Equal to (==):", x == y)               # False
print("Not equal to (!=):", x != y)           # True
print("Greater than (>):", x > y)             # False
print("Less than (<):", x < y)                # True
print("Greater than or equal (>=):", x >= 5)  # True
print("Less than or equal (<=):", y <= 10)    # True
print()
```

```python
# 3. LOGICAL OPERATORS

is_sunny = True
has_umbrella = False

print("--- Logical Operators ---")
# 'and' requires both parts to be True
print("Using 'and':", is_sunny and has_umbrella)  # False

# 'or' requires at least one part to be True
print("Using 'or':", is_sunny or has_umbrella)    # True

# 'not' flips the Boolean value
print("Using 'not':", not is_sunny)                # False
print()
```

```python
# 4. ASSIGNMENT OPERATORS

num = 20
print("--- Assignment Operators ---")

num += 5
print("After += 5:", num)   # 25 (20 + 5)

num -= 3
print("After -= 3:", num)   # 22 (25 - 3)

num *= 2
print("After *= 2:", num)   # 44 (22 * 2)

num /= 4
print("After /= 4:", num)   # 11.0 (44 / 4)
print()
```
